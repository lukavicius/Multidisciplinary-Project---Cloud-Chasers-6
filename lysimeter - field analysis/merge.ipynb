```python
import pandas as pd
from pathlib import Path

# --- paths to your two logger exports ---
file_a = Path("C:\\code\\uni\\mdp\\lysimeter\\lysimeter_evap\\field\\lys2.csv")
file_b = Path("C:\\code\\uni\\mdp\\lysimeter\\lysimeter_evap\\field\\field_log.csv")
out_file = Path("C:\\code\\uni\\mdp\\lysimeter\\lysimeter_evap\\field\\lys2_merged.csv")

dfs = []
for f in (file_a, file_b):
    d = pd.read_csv(f, parse_dates=["timestamp"])
    d["source_file"] = f.name          # keep track of where each row came from
    dfs.append(d)

df = pd.concat(dfs, ignore_index=True)

# Drop rows duplicated across both files (same boot + same elapsed time = same reading)
n_before = len(df)
df = df.drop_duplicates(subset=["boot_session", "elapsed_sec", "timestamp"], keep="first")

# Chronological order
df = df.sort_values(["timestamp", "boot_session", "elapsed_sec"]).reset_index(drop=True)

print(f"{n_before} rows in → {len(df)} rows out ({n_before - len(df)} duplicates removed)")
print(f"Range: {df['timestamp'].min()} → {df['timestamp'].max()}")
print(f"Boot sessions: {sorted(df['boot_session'].unique())}")

df.drop(columns="source_file").to_csv(out_file, index=False)
df.head(10)
```
